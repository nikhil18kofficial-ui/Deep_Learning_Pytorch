```python
import torch
from torch import nn
import matplotlib.pyplot as plt
```

```python
#1.Data Preparing and loading
```

```python
#Creating a general linear regression dataset

m=0.7
c=0.3
x=torch.arange(start=0,end=1,step=0.02).unsqueeze(dim=1)
y_=m*x+c
x[:10],y_[:10]
```

```python
#Splitting data into traing or testing
train_split=int(0.8*len(x))
x_train,y_train=x[:train_split],y_[:train_split]
x_test,y_test= x[train_split:],y_[train_split:]
```

```python
len(x_test),len(y_test),len(x_train),len(y_train)
```

```python
def plot_predictions(train_data=x_train,
                     train_labels=y_train,
                     test_data=x_test,
                     test_labels=y_test,
                     predictions=None):
  """
  Plots training data, test data and compares predictions.
  """
  plt.figure(figsize=(10, 7))

  # Plot training data in blue
  plt.scatter(train_data, train_labels, c="b", s=4, label="Training data")

  # Plot test data in green
  plt.scatter(test_data, test_labels, c="g", s=4, label="Testing data")

  if predictions is not None:
    # Plot the predictions in red (predictions were made on the test data)
    plt.scatter(test_data, predictions, c="r", s=4, label="Predictions")

  # Show the legend
  plt.legend(prop={"size": 14});
```

```python
plot_predictions()
```

```python
#Create linear regression module class
```

```python
import torch
from torch import nn #------>Building blocks to create neural network
class LinearRegressionModel(nn.Module):
  def __init__(self):
    super().__init__()
    self.weight=nn.Parameter(torch.randn(1,requires_grad=True,dtype=torch.float))
    self.bias=nn.Parameter(torch.randn(1,requires_grad=True,dtype=torch.float))
  def forward(self,x:torch.tensor)->torch.tensor:
    return self.weight*x+self.bias
```

```python
#We can know out model parameters
#Create random seed
torch.manual_seed(42)
model_0=LinearRegressionModel()
list(model_0.parameters())
```

```python
with torch.inference_mode():
  y_preds=model_0(x_test)
y_preds
```

```python
y_test
```

```python
plot_predictions(predictions=y_preds)
```

```python
y_pred=model_0(x_test)
y_pred
```

```python
#Training model with pytorch
```

```python
list(model_0.parameters())
```

```python
model_0.state_dict()
```

```python
#Set up Loss function
loss_fn=nn.L1Loss()
#For optimizer
optimizer=torch.optim.SGD(params=model_0.parameters(),lr=0.01)
```

```python
#Building testing loop with training loop
#Step 0: Epoch
epoch_count=[]
train_loss_values=[]
test_loss_values=[]
epochs=100
for epoch in range(epochs):
  #Set model in training mode
  model_0.train()
  #Forward pass
  y_pred=model_0(x_train)
  #Calculate loss
  loss=loss_fn(y_pred,y_train)
  #Optimizer zero grad
  optimizer.zero_grad()
  #Perform Backpropogation wrt to parameters of the model
  loss.backward()
  #Step the optimizer
  optimizer.step() #By default how the optimizer change will accumulate through the loop so..We have to zero them in step 3 for the next iteration of the loop


  #Testing
  #Set model in eval mode
  model_0.eval()
  with torch.inference_mode():
    test_pred=model_0(x_test)
    test_loss=loss_fn(test_pred,y_test)
  if epoch%10==0:
    epoch_count.append(epoch)
    train_loss_values.append(loss)
    test_loss_values.append(test_loss)
  print(f"Epoch:{epoch},Loss:{loss},test_loss:{test_loss}")
```

```python
#Building Pytorch Linear model
```

```python
import torch
from torch import nn
class LinearRegressionModelV2(nn.Module):
  def __init__(self):
    super().__init__()
    self.linear_layer=nn.Linear(in_features=1,out_features=1)
  def forward(self,x:torch.tensor) -> torch.tensor:
    return self.linear_layer(x)
```

```python
torch.manual_seed(42)
model_1=LinearRegressionModelV2()
model_1,model_1.state_dict()
```

```python
#Training a model
```

```python
#Check the models current device
```

```python
next(model_1.parameters()).device
```

```python
device="cuda" if torch.cuda.is_available() else "cpu"
model_1.to(device)
```

```python
next(model_1.parameters()).device
```

```python
#Training code
#Set up loss function and optimizer
```

```python
loss_fn=nn.L1Loss()
optimizer=torch.optim.SGD(params=model_1.parameters(),lr=0.01)
```

```python
#Trining Loop
torch.manual_seed(42)
epochs=200
#We should change the data also to gpu
x_train=x_train.to(device)
y_train=y_train.to(device)
x_test=x_test.to(device)
y_test=y_test.to(device)
for epoch in range(epochs):
  model_1.train()
  #Forward pass
  y_pred=model_1(x_train)
  #Calculate the loss
  loss=loss_fn(y_pred,y_train)
  #Optimizer zero grad
  optimizer.zero_grad()
  #Perform BackPropogation
  loss.backward()
  #Calculate the gradients
  optimizer.step()

  ##Testing Loop
  model_1.eval()
  with torch.inference_mode():
    test_pred=model_1(x_test)
    test_loss=loss_fn(test_pred,y_test)
  #Print out whats happ
  if epoch%10==0:
    print(f"Epoch: {epoch} , Loss: {loss} , Test Loss: {test_loss}")
```

```python
#Making Predictions on test data
model_1.eval()
with torch.inference_mode():
  y_preds=model_1(x_test)
y_preds
```
